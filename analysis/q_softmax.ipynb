```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, Layout

def solve_softmax_action(c_left, c_right, c_wait, inv_temp):
    # --- 1. Your Logic ---
    q_vals = np.array([c_left, c_right, c_wait])
    q_vals = -q_vals                # Convert cost to value (negate)
    q_vals -= np.max(q_vals)        # Stability shift
    
    # Apply temperature
    # We add a tiny epsilon to inv_temp to avoid strictly 0 multiplication issues if needed
    probs = np.exp(inv_temp * q_vals)
    probs /= np.sum(probs)
    
    # --- 2. Visualization ---
    actions = ['Left', 'Right', 'Wait']
    colors = ['#e74c3c', '#3498db', '#f1c40f'] # Red, Blue, Yellow
    
    plt.figure(figsize=(8, 4))
    
    # Bar Chart
    bars = plt.bar(actions, probs, color=colors, alpha=0.8)
    
    # Styling
    plt.ylim(0, 1.1)
    plt.ylabel('Probability')
    plt.title(f'Softmax Selection (Inv Temp: {inv_temp:.1f})')
    plt.grid(axis='y', linestyle='--', alpha=0.3)
    
    # Add text labels on bars
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 0.02,
                 f'{height:.2f}', ha='center', fontweight='bold')
    
    # Also print the raw values for clarity
    print(f"Input Costs: {actions} = [{c_left}, {c_right}, {c_wait}]")
    print(f"Probabilities: {np.round(probs, 3)}")
    plt.show()

# --- 3. Create Interactive Sliders ---
# This automatically generates the UI controls
interact(
    solve_softmax_action, 
    c_left=FloatSlider(value=1.0, min=0, max=2, step=0.01, description='Cost Left'),
    c_right=FloatSlider(value=1.0, min=0, max=2, step=0.01, description='Cost Right'),
    c_wait=FloatSlider(value=1.0, min=0, max=2, step=0.01, description='Cost Wait'),
    inv_temp=FloatSlider(value=50.0, min=1, max=100, step=1, description='Inv Temp (β)')
);
```
